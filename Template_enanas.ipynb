```python
from astropy import constants as c
import numpy as np
import matplotlib.pyplot as plt
import random
import seaborn as sns
from scipy.stats import norm
```

```python
def IMF(m):
    if m >= 0.01 and m < 0.08:
        alpha = 0.3
        c=1
    elif m >= 0.08 and m < 0.50:
        alpha = 1.3
        c=1
    elif m >= 0.50 and m < 1:
        alpha = 2.3
        c=0.5
    elif m >= 1:
        c=0.5
        alpha = 2.3

    xi = c*m**(-alpha) 

    return xi
```

```python
x = np.arange (0.08, 15, 0.01)
```

```python
plt.plot(x, [IMF(i)/IMF(0.08) for i in x])
```

```python
numeroloco = random.uniform(0.08,15)
print(numeroloco)
print(IMF(numeroloco)/IMF(0.08))
unoysero = random.uniform(0,1)
print(unoysero)
```

```python
#Ahora toca revisar si el número aleatorio que generé cae o no cae en el coso 
```

```python
contador = 0
guardo = []
while contador < 100000:
    numeroloco = random.uniform(0.08,15) 
    unoysero = random.uniform(0,1)
    nono = IMF(numeroloco)/IMF(0.08)
    if nono >= unoysero:
        guardo.append(numeroloco)
        contador +=1
```

```python
#Que el star formation rate sea constante significa que se forman estrellas a la misma tasa a través del tiempo
# Ahora tenemos que evolucionarlas
# Además que tengamos el mismo star formation rate, significa que cualquiera de las masas que ya simulamos se pueden
# crear en casi todos los tiempos de la galaxia

# Por esto mismo deberíamos ahora encontrar un star formation rate en función del tiempo debido a los distintos procesos
# Age metallicity relation, pensando en las supernovas. 
# IMF, ya hecho
# SFR, que dijimos 
# Age-z, o sea la metalicidad con respecto del coso. 
```
